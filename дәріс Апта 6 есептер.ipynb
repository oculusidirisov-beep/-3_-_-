```python
#1. Series — негізгі тапсырмалар
# 1. Pandas кітапханасын импорттаймыз
import pandas as pd

# 2. temperatures деген Series құрамыз
temperatures = pd.Series({
    'Алматы': 5,
    'Астана': -3,
    'Шымкент': 7
})

# 3. Series-ті экранға шығару
print("Барлық қалалар мен олардың температурасы:")
print(temperatures)

# 4. Алматының температурасын алу
almaty_temp = temperatures['Алматы']
print("\nАлматының температурасы:", almaty_temp)

# 5. Температурасы 0-ден жоғары қалаларды сүзу
positive_temps = temperatures[temperatures > 0]
print("\nТемпературасы 0-ден жоғары қалалар:")
print(positive_temps)
```

```python
#2. DataFrame — кестемен жұмыс
import pandas as pd

# 1. Студенттер туралы DataFrame құрамыз
data = {
    'name': ['Aigerim', 'Dias', 'Miras', 'Dana'],
    'age': [20, 22, 21, 19],
    'score': [88, 92, 75, 95]
}

students = pd.DataFrame(data)

# 2. DataFrame-ті экранға шығару
print("Барлық студенттер:")
print(students)

# 3. Жасы 20-дан жоғары студенттерді сүзу
older_than_20 = students[students['age'] > 20]
print("\nЖасы 20-дан жоғары студенттер:")
print(older_than_20)

# 4. Тек name және score бағандарын шығару
name_score = students[['name', 'score']]
print("\nСтуденттердің аты мен балы:")
print(name_score)

# 5. Ең жоғары балл жинаған студенттің атын табу
top_student = students.loc[students['score'].idxmax(), 'name']
print("\nЕң жоғары балл жинаған студент:", top_student)
```

```python
#3. Индексация және таңдау iloc арқылы 3-жолдың 2-бағанын алыңыз.
#loc көмегімен студенттің аты “Dias” болған жолды шығарыңыз.
import pandas as pd

# Студенттер туралы DataFrame
data = {
    'name': ['Aigerim', 'Dias', 'Miras', 'Dana'],
    'age': [20, 22, 21, 19],
    'score': [88, 92, 75, 95]
}

students = pd.DataFrame(data)
print("Барлық студенттер:")
print(students)
#iloc арқылы таңдау
#iloc — индекс бойынша таңдау (сандық индекстер, 0-ден басталады).
value = students.iloc[2, 1]
print("3-жолдың 2-баған мәні:", value)

#loc арқылы таңдау
#loc — индекс немесе шарт бойынша таңдау (сандық емес, нақты мәндерге негізделген).
dias_row = students.loc[students['name'] == 'Dias']
print("\nСтуденттің аты Dias:")
print(dias_row)

```

```python
#5. Groupby — топтастыру
#Қызметкерлер кестесін жасаңыз:
import pandas as pd

# 1. Қызметкерлер туралы DataFrame құрамыз
data = {
    'name': ['Ayan', 'Dana', 'Miras', 'Aida', 'Timur'],
    'department': ['IT', 'HR', 'IT', 'HR', 'Sales'],
    'salary': [400000, 300000, 450000, 320000, 280000]
}

employees = pd.DataFrame(data)

# 2. Әр бөлім бойынша орташа жалақыны есептеу
avg_salary = employees.groupby('department')['salary'].mean()
print("Әр бөлім бойынша орташа жалақы:")
print(avg_salary)

# 3. Әр бөлімдегі қызметкерлер санын шығару
count_per_dept = employees.groupby('department')['name'].count()
print("\nӘр бөлімдегі қызметкерлер саны:")
print(count_per_dept)

# 4. Ең жоғары жалақысы бар бөлімді анықтау
max_salary_dept = employees.groupby('department')['salary'].max().idxmax()
print("\nЕң жоғары жалақысы бар бөлім:", max_salary_dept)
```

```python
#6. Агрегаттау (sum, mean, max, min)
import pandas as pd

# 5-тапсырмадағы қызметкерлер кестесі
df = pd.DataFrame({
    'name': ['Ayan', 'Dana', 'Miras', 'Aida', 'Timur'],
    'department': ['IT', 'HR', 'IT', 'HR', 'Sales'],
    'salary': [400000, 300000, 450000, 320000, 280000]
})

# 1) Әр бөлім үшін жалақының ең кішісі мен ең жоғарғысы
dept_min_max = df.groupby('department')['salary'].agg(['min', 'max'])
print("Әр бөлім бойынша жалақының ең кішісі және ең жоғарғысы:")
print(dept_min_max, "\n")

# 2) Әр бөлім бойынша жалақының суммасын есептеу
dept_sum = df.groupby('department')['salary'].sum()
print("Әр бөлім бойынша жалақы суммаcы:")
print(dept_sum)

```

```python
import pandas as pd

# Берілген кесте
df = pd.DataFrame({
    'student': ['Aigerim', 'Dias', 'Ayan'],
    'math': [88, 75, 95],
    'physics': [90, 80, 92],
    'history': [78, 85, 90]
})

# 1) Әр пәннен жалпы статистика: min, max, mean, median
subject_stats = df[['math', 'physics', 'history']].agg(['min', 'max', 'mean', 'median'])
print("1) Пәндер бойынша статистика (min, max, mean, median):")
print(subject_stats, "\n")

# 2) Әр студенттің орташа балын есептеу
df['avg'] = df[['math', 'physics', 'history']].mean(axis=1)
print("2) Орташа бал қосылған кесте:")
print(df, "\n")

# 3) Орташа балы 85-тен жоғары студенттер
high_avg = df[df['avg'] > 85]
print("3) Орташа балы 85-тен жоғары студенттер:")
print(high_avg)
```
