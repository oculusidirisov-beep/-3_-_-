```python
from transformers import AutoTokenizer

sentence = "Natural Language Processing is amazing and fun to learn."

# 1. 'bert-base-uncased' моделінің токенизаторын жүктеңіз
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

# 2. sentence айнымалысын токенизатор арқылы өңдеп, нәтижені 'inputs' айнымалысына сақтаңыз
inputs = tokenizer(sentence)

# 3. Нәтижені экранға шығарыңыз
print(inputs)
```

```python
from transformers import pipeline

texts = [
    "I love using Python for data science!",
    "The weather is terrible today, I hate it."
]

# 1. 'sentiment-analysis' үшін pipeline құрыңыз
classifier = pipeline("sentiment-analysis")

# 2. 'texts' тізімін pipeline-ға беріп, нәтижесін алыңыз
results = classifier(texts)

# 3. Әр мәтіннің label (белгісі) мен score (сенімділік деңгейін) экранға шығарыңыз
for text, result in zip(texts, results):
    print(f"Мәтін: {text}")
    print(f"Белгі: {result['label']}, Сенімділік: {result['score']:.4f}\n")
```

```python
from transformers import pipeline

ner_text = "Elon Musk is the CEO of SpaceX, located in California."

# 1. NER тапсырмасы үшін pipeline құрыңыз
ner_pipe = pipeline("ner", aggregation_strategy="simple")

# 2. 'ner_text' айнымалысын модельге беріңіз
entities = ner_pipe(ner_text)

# 3. Табылған нысандарды экранға шығарыңыз
for ent in entities:
    print(f"Entity: {ent['entity_group']}, Score: {ent['score']:.4f}, Word: {ent['word']}")
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

long_text = """
The Transformer model architecture, introduced in the paper "Attention Is All You Need",
has revolutionized the field of Natural Language Processing. Unlike previous Recurrent Neural Networks (RNNs),
Transformers process the entire input sequence in parallel, allowing for much faster training and
better handling of long-range dependencies in text. This architecture is the foundation of
modern models like BERT, GPT, and T5.
"""

# 1. Токенизатор мен модельді жүктеңіз
tokenizer = AutoTokenizer.from_pretrained("facebook/bart-large-cnn")
model = AutoModelForSeq2SeqLM.from_pretrained("facebook/bart-large-cnn")

# 2. Мәтінді токенизациялаңыз
inputs = tokenizer(long_text, return_tensors="pt", max_length=1024, truncation=True)

# 3. Мәтінді қысқартыңыз
summary_ids = model.generate(
    inputs["input_ids"],
    max_length=50,
    min_length=20,
    num_beams=4, # Use beam search for better quality
    early_stopping=True
)

# 4. Қысқартылған мәтінді экранға шығарыңыз
summary_text = tokenizer.decode(summary_ids[0], skip_special_tokens=True)
print(summary_text)
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

text_to_translate = "Hugging Face makes deep learning accessible to everyone."

# 1. Токенизатор мен модельді жүктеңіз
tokenizer = AutoTokenizer.from_pretrained("Helsinki-NLP/opus-mt-en-fr")
model = AutoModelForSeq2SeqLM.from_pretrained("Helsinki-NLP/opus-mt-en-fr")

# 2. Мәтінді токенизациялаңыз
inputs = tokenizer(text_to_translate, return_tensors="pt")

# 3. Мәтінді аударыңыз
translated_ids = model.generate(inputs["input_ids"])

# 4. Аударма нәтижесін көрсетіңіз
translated_text = tokenizer.decode(translated_ids[0], skip_special_tokens=True)
print(translated_text)
```

```python
from transformers import pipeline

model_name = "xlm-roberta-base"
sentence = "Қазақстанның астанасы <mask> қаласы" # xlm-roberta үшін <mask> қолданылады

# 1. 'fill-mask' үшін pipeline құрыңыз
mask_filler = pipeline("fill-mask", model=model_name)

# 2. Нәтижені алыңыз
results = mask_filler(sentence)

# 3. Ең ықтимал сөзді шығарыңыз
top_result = results[0]
print(f"Болжам: {top_result['token_str']}, Ұпай: {top_result['score']:.4f}")
```

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification, TrainingArguments

model_name = "xlm-roberta-base"

# 1. Токенизаторды жүктеңіз
tokenizer = AutoTokenizer.from_pretrained(model_name)

# 2. Модельді жүктеңіз (2 класс үшін)
model = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=2)

# 3. TrainingArguments объектісін құрыңыз
training_args = TrainingArguments(
    output_dir="./kaz_bert_finetuned",
    learning_rate=2e-5,
    num_train_epochs=3,
    weight_decay=0.01,
)

# 4. Аргументтерді тексеру
print(training_args)
```

Қорытынды сұрақтарға жауаптар:
Attention Mask не үшін қажет?

Бұл модельге қай токендер нақты мәтін екенін, ал қайсысы тек ұзындықты теңестіру үшін қосылған толтырғыштар (padding) екенін түсінуге көмектеседі. Модель назарын (attention) тек маңызды сөздерге аударады.

Extractive және Abstractive саммаризацияның айырмашылығы неде?

Extractive — мәтіндегі бар сөйлемдерді сол күйінде таңдап алып, "қиып-жапсыру".

Abstractive — мәтіннің мағынасын түсініп, оны өз сөзімен, жаңа сөйлемдермен қысқартып жазу (адам сияқты).

Неліктен қазақ тілі үшін арнайы модельді қолданған дұрыс?

Себебі арнайы модельдер қазақ тілінің грамматикасын, морфологиясын және контекстін жақсы түсінеді. bert-base-uncased негізінен ағылшын тілінде оқытылғандықтан, қазақша сөздерді дұрыс танымауы немесе мағынасыз бөліктерге бөліп тастауы мүмкін.
